```python
class Error(Exception):
    pass
class dobException(Error):
    pass
```

```python
year = int(input("Enter your year of birth: "))
age = 2025 - year

try:
    if age <= 30 and age >= 20:
        print("Your age is valid, you can apply for the exams.")
    else:
        raise dobException
except dobException as e:
    print("Your age is not valid, you cannot apply for the exams.")
```
