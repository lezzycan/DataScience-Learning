```python
# Universal functions (ufuncs) for element-wise operations


arr = np.array([4, 9, 16, 25, 36])
print(np.sqrt(arr))
print(np.exp(arr))  # Exponential of each element in the array
print(np.log(arr)) 
print(np.sin(arr))# Natural logarithm of each element in the array
# Square root of each element in the array
```

```python
# statistcal concepts --- Normalization

def normalize(arr):
    """Normalize a numpy array to have mean 0 and standard deviation 1."""
    mean = np.mean(arr)
    std = np.std(arr)
    return (arr - mean) / std
```

```python
normalized_arr = normalize([1,2,3,4,5])
print("Normalized array:", normalized_arr)
```
