```python
# filling missing values with 0
data_filled = data.fillna(0)  # Fill null values with 0
data_filled.isnull().sum()  # Check again for null values
```

```python
data_filled_mean = data.fillna(data.select_dtypes(include='number').mean())  # Fill null values with the mean of each column 
data_filled_mean
```

```python
data.isnull().sum()
```

```python
data_filled
```

```python
data_filled_mean['Events'].fillna(data_filled_mean['Events'].mode()[0])  # Fill null values with the mode of the 'Events' column
```

```python
data_filled_mean.head(5)
```

```python
data_filled_mean = data_filled_mean.rename(columns = {"Date": "Temp Data"})
data_filled_mean.head(5)
```

```python
grouped_mean = data_filled_mean.groupby('Events')['Max.TemperatureC'].mean()
print(grouped_mean)
```

```python
grouped_agg = data_filled_mean.groupby('Events')['Max.TemperatureC'].agg({ 'mean', 'min', 'max', 'std' })
print(grouped_agg)
```

```python
# Merging and joining DataFrames

data1 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie'],    
    'Age': [25, 30, 35]
})
data2 = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'David'],    
    'City': ['New York', 'Los Angeles', 'Chicago']
})
data1
```

```python
data2
```

```python
#Merge DataFrames on key column
merged_data = pd.merge(data1, data2, on='Name', how='inner')
merged_data
```

```python
pd.merge(data1, data2, on='Name', how='outer')
```

```python
pd.merge(data1, data2, on='Name', how='left')
```

```python
pd.merge(data1, data2, on='Name', how='right')
```
