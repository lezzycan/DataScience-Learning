```python
def welcome():
    return "Welcome to the Python script!"


# Closures is a function inside a fuction

def main_welcome(func):
    
    def sub_welcome():
        print("Welcome to the advance python course")
        func()
        print("Please learn these concept properly")
    return sub_welcome()    
```

```python
main_welcome(print)
```

```python
@main_welcome
def course_intro():
    print("This is an advanced python course")
```

```python
def my_decorator(func):
    def wrapper():
        print("Before the function call")
        func()
        print("After the function call")
    return wrapper
```

```python
@my_decorator
def say_hello():
    print("Hello, World!")
```

```python
say_hello()
```

```python
def repeat(num_times):
    def decorator_repeat(func):
        def wrapper(*args, **kwargs):
            for _ in range(num_times):
                func(*args, **kwargs)
        return wrapper
    return decorator_repeat
```

```python
@repeat(3)
def greet():
    print(f"Hello")
```

```python
greet()
```
