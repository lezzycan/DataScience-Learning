```python
try:
    num = int(input("Enter a number: "))
    result = 10 / num
except ValueError as e:
    print ("This is a value error:", e)
except ZeroDivisionError as e:
    print(e)    
except Exception as ex2:
    print(f"Error: {ex2}")
else:
    print(f"Result is: {result}")  
finally:
    print("Execution completed.")          
    
```
